```python
from molzen.io import Molecule

mol = Molecule.from_pubchem(1174)  # uracil
print(f"{len(mol.atom_records)} atoms, {len(mol.bonds)} bonds")
editor = mol.show(edit=True, width=760, height=420)
display(editor)
```

1. Select an oxygen by clicking it or using the atom list. Set **Element** to S and click **Change element**, then **Undo**.
2. Select one hydrogen and click **Replace H with methyl**. Expect 15 atoms and 15 bonds. Set a torsion before placement, or undo and retry another torsion.
3. Try **Undo** and **Redo**. Select two atoms to inspect or change their bond.

The GUI edits a copy. It provides starting geometry, without minimization or automatic charge/spin assignment.

```python
edited = editor.molecule
print("Original:", len(mol.atom_records), "atoms")
print("Edited:", len(edited.atom_records), "atoms")
edited.bonds.records
```

Run the following cell when you want to save. PDB retains connectivity but loses bond orders; MOL2 retains supported bond types; NPY retains the full molzen data. Re-run `edited = editor.molecule` after subsequent GUI changes.

```python
edited = editor.molecule
edited.to_pdb("uracil_edited.pdb")
edited.to_mol2("uracil_edited.mol2")
edited.to_npy("uracil_edited.npy")
```

To experiment with inference, construct a coordinate-only copy below. Click **Suggest bonds**, review the gold candidates, then **Accept suggestions**. Their bond orders remain unknown. Confirm a hydrogen-parent bond using **Set bond → Single** before trying methyl placement.

```python
coordinates_only = Molecule(xyz=mol.xyz.copy(), elements=mol.elements)
inference_editor = coordinates_only.show(edit=True, width=760, height=420)
display(inference_editor)
```
